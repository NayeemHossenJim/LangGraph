```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class Temp_Converter(TypedDict):
    celsius: float
    fahrenheit: float
```

```python
def converter(state: Temp_Converter) -> Temp_Converter:
    c = state["celsius"]
    state["fahrenheit"] = c * 9 / 5 + 32
    return state
```

```python
graph = StateGraph(Temp_Converter)

graph.add_node("temp_converter", converter)

graph.add_edge(START, "temp_converter")
graph.add_edge("temp_converter", END)
```

```python
workflow = graph.compile()
```

```python
final_state = workflow.invoke({"celsius": 100})
final_state
```
